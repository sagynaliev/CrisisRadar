```python
# Sentiment Analysis — жаңалықтардың көңіл-күйін талдау
from textblob import TextBlob

pip_install = __import__('subprocess').run(['pip', 'install', 'textblob'], capture_output=True)

def analyze_sentiment(text):
    if not text:
        return 0, 'Neutral'
    analysis = TextBlob(str(text))
    score = analysis.sentiment.polarity
    if score < -0.1:
        return score, '🔴 Negative'
    elif score > 0.1:
        return score, '🟢 Positive'
    else:
        return score, '🟡 Neutral'

df_news[['sentiment_score', 'sentiment']] = df_news['title'].apply(
    lambda x: pd.Series(analyze_sentiment(x))
)

print("✅ Sentiment анализ дайын!")
print(df_news[['title', 'sentiment_score', 'sentiment']].head(10))
```

```python
# Sentiment визуализация
sentiment_counts = df_news['sentiment'].value_counts().reset_index()
sentiment_counts.columns = ['sentiment', 'count']

fig3 = px.pie(
    sentiment_counts,
    values='count',
    names='sentiment',
    title='📰 Жаһандық жаңалықтар сентименті',
    color_discrete_map={
        '🔴 Negative': '#ff4444',
        '🟡 Neutral': '#ffaa00',
        '🟢 Positive': '#00ff88'
    }
)

fig3.update_layout(
    paper_bgcolor='#0a0a0a',
    font=dict(color='white')
)

fig3.show()

# Timeline
fig4 = px.scatter(
    df_news,
    x='published',
    y='sentiment_score',
    color='sentiment',
    hover_name='title',
    title='📈 Сентимент уақыт бойынша',
    color_discrete_map={
        '🔴 Negative': '#ff4444',
        '🟡 Neutral': '#ffaa00',
        '🟢 Positive': '#00ff88'
    }
)

fig4.update_layout(
    paper_bgcolor='#0a0a0a',
    plot_bgcolor='#1a1a2e',
    font=dict(color='white')
)

fig4.show()
```

```python
import pandas as pd
import numpy as np

# Earthquake деректерін жүктеу
df_clean = pd.read_csv('earthquake_data.csv')
df_clean['time'] = pd.to_datetime(df_clean['time'])
df_clean = df_clean[['time', 'latitude', 'longitude', 'depth', 'mag', 'place']].dropna()

print(f"✅ Earthquake деректері жүктелді: {len(df_clean)} жазба")
```

```python
# Composite Risk Score — барлық қауіптерді біріктіру
# Earthquake + Nuclear + News Sentiment

# 1. Earthquake risk (аймақ бойынша)
eq_risk = df_clean.groupby('country')['mag'].mean().reset_index() if 'country' in df_clean.columns else None

# 2. Nuclear risk
nuclear_risk = df_nuclear[['country', 'risk_score', 'nuclear_warheads', 'stability_index']].copy()

# 3. News sentiment risk
avg_sentiment = df_news['sentiment_score'].mean()
news_risk = 50 + (-avg_sentiment * 100)  # negative news = higher risk

print(f"📰 Жаңалық қауіп деңгейі: {news_risk:.1f}/100")
print(f"☢️ Орташа ядролық қауіп: {df_nuclear['risk_score'].mean():.1f}/100")

# Composite score
composite = pd.DataFrame({
    'country': df_nuclear['country'],
    'nuclear_risk': df_nuclear['risk_score'],
    'stability': df_nuclear['stability_index'],
    'news_risk': news_risk,
    'composite_score': (df_nuclear['risk_score'] * 0.5 + 
                       (10 - df_nuclear['stability_index']) * 5 + 
                       news_risk * 0.2)
}).sort_values('composite_score', ascending=False)

print("\n🌍 Composite Risk Score:")
print(composite[['country', 'composite_score']].to_string(index=False))
```

```python
# Composite Risk Score визуализация
fig5 = px.bar(
    composite.sort_values('composite_score', ascending=True),
    x='composite_score',
    y='country',
    orientation='h',
    color='composite_score',
    color_continuous_scale='RdYlGn_r',
    title='🌍 Global Composite Risk Score (Nuclear + News + Stability)',
    labels={'composite_score': 'Қауіп деңгейі', 'country': 'Ел'}
)

fig5.update_layout(
    paper_bgcolor='#0a0a0a',
    plot_bgcolor='#1a1a2e',
    font=dict(color='white')
)

fig5.show()

# Radar chart — елдер бойынша
fig6 = px.scatter_geo(
    composite.merge(df_nuclear[['country', 'latitude', 'longitude']], on='country'),
    lat='latitude',
    lon='longitude',
    size='composite_score',
    color='composite_score',
    hover_name='country',
    color_continuous_scale='Reds',
    title='🗺️ Global Risk Map — Composite Score',
    projection='natural earth'
)

fig6.update_layout(
    paper_bgcolor='black',
    geo=dict(bgcolor='#0a0a0a', landcolor='#1a1a2e', oceancolor='#16213e'),
    font=dict(color='white')
)

fig6.show()
```

```python
# FINAL COMPOSITE RISK SCORE — барлық қауіптерді біріктіру
# Earthquake + Nuclear + Conflict + EM-DAT + News

final_risk = pd.DataFrame({
    'risk_category': [
        'Nuclear War', 'WW3 / Conflict', 
        'Major Earthquake', 'Climate Disaster',
        'Pandemic', 'Cyber Attack',
        'Tsunami', 'Nuclear Accident'
    ],
    'probability_score': [15, 35, 75, 85, 40, 60, 45, 20],
    'impact_score': [100, 90, 70, 65, 80, 55, 75, 85],
    'preparedness_score': [60, 45, 70, 50, 65, 40, 55, 60]
})

# Composite = probability * impact / preparedness
final_risk['composite'] = (
    final_risk['probability_score'] * 0.4 + 
    final_risk['impact_score'] * 0.4 + 
    (100 - final_risk['preparedness_score']) * 0.2
)

final_risk = final_risk.sort_values('composite', ascending=False)
print("🌍 FINAL GLOBAL RISK ASSESSMENT:")
print(final_risk[['risk_category', 'composite']].to_string(index=False))
```

```python
# Final Risk Dashboard визуализация
fig11 = px.bar(
    final_risk,
    x='composite',
    y='risk_category',
    orientation='h',
    color='composite',
    color_continuous_scale='RdYlGn_r',
    title='🌍 Crisis Radar — Final Global Risk Assessment',
    labels={'composite': 'Composite Risk Score', 'risk_category': 'Қауіп категориясы'},
    text='composite'
)

fig11.update_traces(texttemplate='%{text:.1f}', textposition='outside')

fig11.update_layout(
    paper_bgcolor='#0a0a0a',
    plot_bgcolor='#1a1a2e',
    font=dict(color='white'),
    height=500
)

fig11.show()

# Radar Chart
import plotly.graph_objects as go

categories = final_risk['risk_category'].tolist()
values = final_risk['composite'].tolist()
values.append(values[0])
categories.append(categories[0])

fig12 = go.Figure(go.Scatterpolar(
    r=values,
    theta=categories,
    fill='toself',
    line=dict(color='#ff4444'),
    fillcolor='rgba(255, 68, 68, 0.2)'
))

fig12.update_layout(
    polar=dict(
        bgcolor='#1a1a2e',
        radialaxis=dict(visible=True, range=[0, 100], color='white'),
        angularaxis=dict(color='white')
    ),
    paper_bgcolor='#0a0a0a',
    font=dict(color='white'),
    title='🎯 Global Risk Radar Chart'
)

fig12.show()
```
